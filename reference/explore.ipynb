```python
import requests, socket, urllib3.util.connection
urllib3.util.connection.allowed_gai_family = lambda: socket.AF_INET

url = "https://www.ncei.noaa.gov/pub/data/ghcn/daily/ghcnd-inventory.txt"
inv = requests.get(url, timeout=120).text.splitlines()

print(len(inv))
print(inv[0])
print(inv[1])

```

```python
new_arr = []
params = ["TMAX", "PRCP"]
firstyear=1990
lastyear=2025

for line in inv:
    if line[31:35].strip() in params and int(line[36:40])<=firstyear and int(line[41:45])>=lastyear:
        new_arr.append(line)

print(len(new_arr))
print(new_arr[:5])
```
